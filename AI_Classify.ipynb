```python
%sql
CREATE OR REPLACE TEMP VIEW raw_structured_doc AS 
SELECT
  *
FROM read_files('/Volumes/idp/default/pdf_files')
;
```

```python
%sql
select *
from raw_structured_doc
;
```

```python
%sql
CREATE OR REPLACE TEMP VIEW parsed_structured_doc AS
SELECT 
  *,
  ai_parse_document(content) as parsed_document
FROM raw_structured_doc
;
```

```python
%sql
SELECT 
    path
    ,ai_classify(try_cast(parsed_document AS STRING), ARRAY('Medical Claim', 'Medical Invoice', 'Lab Report', 'Clinical Note', 'Administrative Doc')) as label
FROM parsed_structured_doc
;
```
