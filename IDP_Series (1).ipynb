```python
%sql
-- Parse the PDF content using ai_parse_document and create a structured view
CREATE OR REPLACE TEMP VIEW parsed_structured_doc AS 
SELECT 
  * 
  ,ai_parse_document(content) AS parsed_content
FROM raw_unstructured_doc
;
```

```python
%sql
-- Preview the contents of the parsed_structured_doc view
select * from parsed_structured_doc
```

```python
%sql
-- Extract table elements from parsed_content and create a view with HTML tables
CREATE OR REPLACE TEMP VIEW structured_tables AS 
SELECT 
  path
  ,try_cast(e:content AS STRING) as table_html
FROM parsed_structured_doc
LATERAL VIEW explode(try_cast(parsed_content:document:elements AS ARRAY<VARIANT>)) t AS e
WHERE try_cast(e:type AS STRING) = 'table'
;
```

```python
%sql
-- Preview the contents of the structured_tables view
select * from structured_tables
```

```python
%sql
-- Extract structured rows from HTML tables in table_html
with rows as 
  (
  select *
  from structured_tables
  LATERAL VIEW explode(regexp_extract_all(table_html,'<tr>(.*?)</tr>',1)) t AS tr
  ), data_rows as 
  (
  select path, tr
  from rows
  where tr like '%<td>%'
  )
  select 
    ai_extract(tr, array('CPT Code')).`CPT Code` as CPT_Code,
    ai_extract(tr, array('ICD Code')).`ICD Code` as ICD_Code,
    ai_extract(tr, array('Description')).`Description` as Description,
    ai_extract(tr, array('Billed Amount')).`Billed Amount` as Billed_Amount,
    ai_extract(tr, array('Paid Amount')).`Paid Amount` as Paid_Amount
  from data_rows
```

### Extracting Tables from PDF Documents

This section demonstrates how to extract structured table data from the `table_html` column in the `structured_tables` view. The process uses Python libraries (BeautifulSoup and pandas) to parse HTML tables embedded in PDF files and convert them into DataFrames for further analysis.

**Workflow:**
* Query the `table_html` column from the SQL view.
* Parse each HTML table using BeautifulSoup.
* Convert the parsed table to a pandas DataFrame.
* Combine all tables and display the results.

This approach enables you to analyze tabular data extracted from unstructured PDF documents in a structured format.

```python
# Extract table data from table_html column in structured_tables
# This cell parses HTML tables from the 'table_html' column and converts them to pandas DataFrames for analysis.

from bs4 import BeautifulSoup
import pandas as pd

# Query the table_html column from structured_tables
html_df = spark.sql("SELECT path, table_html FROM structured_tables WHERE table_html IS NOT NULL")

# Initialize a list to store parsed DataFrames
parsed_tables = []

# Iterate through each row and parse the HTML table
for row in html_df.collect():
    # Parse the HTML content using BeautifulSoup
    soup = BeautifulSoup(row['table_html'], 'html.parser')
    # Find the first <table> element in the HTML
    table = soup.find('table')
    if table:
        # Use pandas to read the HTML table into a DataFrame
        df = pd.read_html(str(table))[0]
        # Add the source file path to the DataFrame for traceability
        df['source_path'] = row['path']
        parsed_tables.append(df)

# Combine all parsed tables into a single DataFrame
if parsed_tables:
    all_tables_df = pd.concat(parsed_tables, ignore_index=True)
    # Display the first few rows of the combined DataFrame
    display(all_tables_df.head())
else:
    print('No tables found in table_html column.')
```
